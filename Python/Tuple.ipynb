# Tuples
-  Collection of elements with different data types
-  it is not mutable

```python
# ( )

my_tuple = ("machine", "learning", 1, 2 ,3, 5.5)
print(my_tuple)
```

```python
type(my_tuple)
```

- print first three elements
- print last three elements
- print alternate elements in my_tuple
- print every 3rd element in my_tuple
- print my_tuple in reverse order

```python
# print first three elements
my_tuple[:3]
```

```python
# print last three elements
my_tuple[-3:]
```

```python
# print alternate elements in my_tuple starting with index 2
my_tuple[2::2]
```

```python
# print every 3rd element in my_tuple index 0
my_tuple[::3]
```

```python
# print my_tuple in reverse order
my_tuple[::-1]
```

### NOT MUTABLE

```python
print(my_tuple)
```

```python
my_tuple.count('learning')
```

```python
my_tuple.index(1)
```

```python
my_tuple.index('asdada')
```

## + -> concatenation operator

```python
my_tuple2 = ('data','science')

my_tuple3 = my_tuple + my_tuple2
print(my_tuple3)
```
