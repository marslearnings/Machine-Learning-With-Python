# Sets
- collection of elements with different data types 
- collection unique elements
- indixing cant be perform in set
- Sets follow order 

`set` , `{ }`

```python
A = {1,1,2,3,3,4,5,5,6,7,7}
print(A)
```

```python
B = {9,9,1,5,0,7,7,8,6}
print(B)
```

### Methods

```python
# add -> will add the element in the set if the value doen't exists
A.add('data')
print(A)
```

```python
B.add(8)
print(B)
```

```python
# A - B
# B - A
print("A =", A)
print("B =", B)

print("Difference between A - B = ", A.difference(B))
```

```python
C = A.copy()
print(C)
```

```python
C.difference_update(B)
print(C)
```

```python
# Discard 
# print(A)
A.discard('data') # it will remove and update it
```

```python
print(A)
```

```python
# Intersection
# Common elements between the sets
# A intersection B, B intersection A
A.intersection(B)
```

```python
# Disjoint
# two sets are said to be disjoint if there is no common element between sets
A.isdisjoint(B) # True or False
```

```python
D = {11,12,13}
A.isdisjoint(D)
```

```python
# Subset
# A is called subset of B, if all elements in A are also elements in B
A.issubset(B)
```

```python
print(A)
```

```python
E = {1,2,3}
E.issubset(A)
```

```python
A.issuperset(E)
```

```python
# symmetric_difference
# complete opposite of interection
print(A)
print(B)
```

```python
A.symmetric_difference(B)
```

```python
print(A)
print(B)
print("A union B =")
A.union(B)
```
