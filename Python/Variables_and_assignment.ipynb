### Declaring Variables
Python support the various variable types.
1. int – Integer type.
2. float – Floating point numbers.
3. bool – Booleans are subtypes of integers and assigned value using literals True and False.
4. str – Textual data.

```python
var1 = 2 # integer
var2 = 5.0 # float
v3 = True
```

```python
print("variable - 1= ", var1)
print("variable - 2= ", var2)
print("variable - 3= ", v3)
```

### Type Casting
Converting one data type into another data type
- int to float
- int to str
- float to int
- float to str
- bool to int
- bool to float
- bool to string
- str to int [`conditional`]
- str to float [`conditional`]

```python
int_number = 5
print(type(int_number))
```

```python
# convert int to float
float_number = float(int_number)
print(float_number)
print('data type =', type(float_number))
```

```python
# convert into to str
str_number = str(int_number)
str_float_number = str(float_number)

print(str_number, str_float_number)
print('data type str_number =', type(str_number))
print('data type str_float_number =', type(str_float_number))
```

```python
str_number, str_float_number
```

```python
# Float to interger
val1 = 5.1
val2 = 5.51
# int command will not round off to near value
int_val1 = int(val1)
int_val2 = int(val2)
print('int_val1 =', int_val1)
print('int_val2 =', int_val2)
```

#### Special commands to convert from float to int
1. round
2. ceil - it always round up to near number (next)
3. floor - it always round down to near number (previous)

```python
int_val1 = round(val1)
int_val2 = round(val2)
print('int_val1 =', int_val1)
print('int_val2 =', int_val2)
```

```python
from math import ceil, floor
```

```python
# Ceil rounds a number up to the nearest interger
int_val1 = ceil(val1) # 5.1
int_val2 = ceil(val2) # 5.51
print('int_val1 =', int_val1)
print('int_val2 =', int_val2)
```

```python
# Floor rounds a number down to the nearest interger
int_val1 = floor(val1) # 5.1
int_val2 = floor(val2) # 5.51
print('int_val1 =', int_val1)
print('int_val2 =', int_val2)
```

```python
# bool to str, float, int
# 0 or 1 : False or True
# any value greater 0 and less than 0 then the value is 1 or True
# if the values is zero 0 then the value is 0 or False
bool_val = bool(-1)
print(bool_val)
```
