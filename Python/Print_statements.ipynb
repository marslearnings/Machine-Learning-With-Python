#### Escape Sequence Characters
- \n : new line
- \t : horizontal tab

```python
print("Hello\nworld")
```

```python
print("a","b","c","d",end=' ')
print("Hello world",end=' ')
print('Hi, how are you')
```

```python
# horizontal tab is tab space
print("Hello\t\t\tworld")
```

```python
print("a","b","c","d",sep="\n", end='\t')
print("Hello world",end='\t')
print('Hi, how are you')
```

- \ is a escape character

```python
print('\\n') # escape the sequence
```

```python
print("c:\\notebook\\tables")
```

- \'
- \"

```python
print("This is in double quote")
print('This is in single quote')
```

#### Saving values using print

```python
f = open("print_file.txt",mode='a') # w = write, a = append
print(2+3,file=f)
f.close() # this is important 
# always the file after execution
```
