````python
var1 = input(prompt="Enter a value: ")
# condition for doing data validate

# check if the var1 is numeric - int
# if this condition
# replace . with '' empty and check the result is numeric - int
numeric_cond = var1.isnumeric()
var2 = var1.replace('.','')
float_cond = var2.isnumeric()

if numeric_cond:
    print('Value is integer')
    var1 = float(var1)

elif float_cond:
    print("Value is float")
    var1 = float(var1)
    
else:
    print('Value is string')

print('Type of the var1 =', type(var1))

# convert data type of var1 to int
# type cast (int)

data_condition = isinstance(var1,str) # use to compare the data type
if data_condition: # true the dataype of  var1 is string
    print('The var1 is of data type string, hence comparision is not possible')
else:
    print('The var1 is numeric hence comparision is possible')
    
    condition = var1 > 1
    if condition: # true
        # code block - 1 tab space : indentation
        print('value is greater than 1')
    else: #if condition is false
        print('value is not greater than 1')
````
