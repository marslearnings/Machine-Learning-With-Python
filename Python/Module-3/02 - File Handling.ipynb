### create csv file

````python
text1 = 'Name,Age,Designation'
text2 = 'Srikanth,27,DS'
text3 = 'RaghuRam,25,DS'
text4 = 'Pritvi,24,JDS'
emp = [text1,text2,text3,text4]
````

````python
emp
````

````python
with open('./Datasets/emp.csv','w') as f:
    for i in emp:
        f.write(i+'\n')
        
    f.close()
````

### Append 
`mode = 'a'`

````python
emp_new = ['Avinash,24,DA','Mahesh,25,DA','Rahul,21,JDA']
````

````python
with open('./Datasets/emp.csv','a') as f:
    for i in emp_new:
        f.write(i+'\n')
        
    f.close()
````
