# File Handling

`open`

> mode of operations
```
mode = 'r' -> read
mode = 'w' -> write
mode = 'a' -> append
```

````python
with open('./dataset/filte.txt',mode ='r') as f:
    text =f.readlines()
    f.close()
````

````python
# write file
with open('srikanth.txt',mode ='w') as f:
    text = 'Innomatics Technology Hub'
    f.write(text+'\n')
    f.close()
````

**write a program to create a comma seperated values (csv) file which is in below form**

| Name  |  Age  | Designation  |
|---|---|---|
| Srikanth  | 27  |  DS  |
| RaghuRam  | 25  | DSC  |
| Ramya  | 26  | JDS  |

using `append`

````python
import os
````

````python
col = 'Name,Age,Designation' 
emp1 = 'Srikanth,25,DS'
emp2 = 'Raghu,25,DSC'
emp3 = 'Ramya,26,JDS'
employee = [col,emp1,emp2,emp3]
directory = './data/'
name = 'emp.csv'

for row in employee:
    files = os.listdir(directory)
    if name in files:
        with open(directory+name,'a')  as f:
            f.write(row+'\n')
            f.close()
    else:
        with open(directory+name,'w')  as f:
            f.write(row+'\n')
            f.close()
````
